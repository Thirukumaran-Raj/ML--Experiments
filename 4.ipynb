```python
import numpy as np

class Perceptron:
    def __init__(self, learning_rate=0.1, epochs=100):
        self.learning_rate = learning_rate
        self.epochs = epochs
        self.weights = None
        self.bias = 0.0

    def fit(self, X, y):
        """Train on features X and labels y, where labels are 0 or 1."""
        X = np.asarray(X, dtype=float)
        y = np.asarray(y, dtype=int)

        self.weights = np.zeros(X.shape[1])
        self.bias = 0.0

        for _ in range(self.epochs):
            for features, label in zip(X, y):
                prediction = int(np.dot(features, self.weights) + self.bias >= 0)
                error = label - prediction
                self.weights += self.learning_rate * error * features
                self.bias += self.learning_rate * error

        return self

    def predict(self, X):
        X = np.asarray(X, dtype=float)
        return (np.dot(X, self.weights) + self.bias >= 0).astype(int)


# Example: learn the AND function
X = [[0, 0], [0, 1], [1, 0], [1, 1]]
y = [0, 0, 0, 1]

model = Perceptron(learning_rate=0.1, epochs=20).fit(X, y)
print(model.predict(X))
```
