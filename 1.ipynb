```python
"Write program to implement univariate regression, bivariate regression and multivariate regression."
```

```python
import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score

def fit_regression(X, y):
    """Fit linear regression and report predictions and metrics."""
    X = np.asarray(X, dtype=float)
    y = np.asarray(y, dtype=float)

    if X.ndim == 1:
        X = X.reshape(-1, 1)
    if y.ndim != 1:
        raise ValueError("y must be a one-dimensional target array.")
    if len(X) != len(y):
        raise ValueError("X and y must contain the same number of samples.")

    model = LinearRegression()
    model.fit(X, y)
    predictions = model.predict(X)

    return {
        "model": model,
        "coefficients": model.coef_,
        "intercept": model.intercept_,
        "mse": mean_squared_error(y, predictions),
        "r2": r2_score(y, predictions),
        "predictions": predictions,
    }


def univariate_regression(x, y):
    """Regression using one predictor."""
    return fit_regression(x, y)


def bivariate_regression(x, y):
    """Regression examining one predictor and one target."""
    return fit_regression(x, y)


def multivariate_regression(X, y):
    """Regression using multiple predictors."""
    X = np.asarray(X, dtype=float)
    if X.ndim != 2 or X.shape[1] < 2:
        raise ValueError("X must contain at least two predictor columns.")
    return fit_regression(X, y)


def show_results(name, results):
    print(f"\n{name}")
    print(f"Intercept: {results['intercept']:.3f}")
    print(f"Coefficients: {results['coefficients']}")
    print(f"MSE: {results['mse']:.3f}")
    print(f"R²: {results['r2']:.3f}")


# Example: one predictor
x = np.array([1, 2, 3, 4, 5])
y = np.array([3, 5, 7, 9, 11])

show_results("Univariate regression", univariate_regression(x, y))
show_results("Bivariate regression", bivariate_regression(x, y))

# Example: multiple predictors
X = np.array([
    [1, 10],
    [2, 20],
    [3, 30],
    [4, 40],
    [5, 50],
])
target = np.array([12, 24, 36, 48, 60])

show_results("Multivariate regression", multivariate_regression(X, target))
```
